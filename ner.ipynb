## Named Entity Recognition (NER) with spaCy

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
text = "Apple Inc. was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne in 1976. Its headquarters are in Cupertino, California. The company released the first iPhone in 2007."
doc = nlp(text)

print("Detected Entities:")
for ent in doc.ents:
    print(f"  - {ent.text} ({ent.label_})")
```

```python
from spacy import displacy

displacy.render(doc, style='ent', jupyter=True)
```
